3. Representing one Nkwen Traders product

```python
Product_name = "Vegetable Oil 5L"
print(Product_name)
Product_price = 7565.0
print(Product_price)
Quantity_in_stock = 100
print(Quantity_in_stock)
```

```python
Total_value = Product_price * Quantity_in_stock
print(f"Total_value:{Total_value}FCFA")
```

```python
print(f"One unit of {Product_name} costs {Product_price}FCFA, a total of {Quantity_in_stock} units are in stock, amounting to a total value of {Total_value}FCFA" )
```
