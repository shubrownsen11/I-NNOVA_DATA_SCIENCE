3.Tuples

.Illustrating that tuples  do not support assigning and are immutable

```python
Numbers = (1,2,3,4,5,6,7,8)
Numbers[1] = 5
```

A tuple can be used to represent never changing data for example, if we want to store days of the week, the tuple would consist of monday, tuesday, wednesday, thursday, friday, saturday and sunday. An attempt to remove one of these days or add any would return an error

```python
Daily_sales = [34928.0,8220.0,3267.0,7683.0,5490.0,35760.0,11666.0,149656.0,20910.0,6248.0]
Daily_sales
```

For loops

```python
Values = [1,2,3,4,5,6,7]
sum = 0
for value in Values:
    sum += value
print(sum)    
```

```python
#Summing the Daily_sales using a for loop
Total_Sales = 0
for sale in Daily_sales:
    Total_Sales += sale
print(f"Total Sales:{Total_Sales}FCFA")    
```

```python
#Calculating the average sale
Average_sale = Total_Sales/len(Daily_sales)
print(f"Average Sale:{Average_sale}FCFA")
```

```python
#Getting the highest value from the Daily sales list
Highest_So_Far = 34928.0
for sale in Daily_sales:
    if sale > Highest_So_Far:
        Highest_So_Far = sale
print(f"Highest Value:{Highest_So_Far}:FCFA")        
```

Summary

```python
print(f"Total Sales:{Total_Sales}FCFA") 
print(f"Average Sale:{Average_sale}FCFA")
print(f"Highest Sales Day:{Highest_So_Far}:FCFA")
```
