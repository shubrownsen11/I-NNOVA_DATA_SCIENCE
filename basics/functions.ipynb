Wrapping the code to classify stock in a function

```python
#wrapping the code to classify stock in a function
def classify_stock(quantity):
    if (quantity == 0):
      return("Out of stock")
    elif(0 < quantity <= 10):
      return("Low Stock")
    else:
      return("Well Stocked")   

#Calling the function in a for loop, to classify sample quantities
Sample_Quantities = [24, 1, 0, 11, 7, 14, 2]
for Quantity in Sample_Quantities:
   print(f"{Quantity}:",classify_stock(Quantity))
```

Wrapping the code to calculate inventory value in a function

```python
def calculate_inventory_value(Products):
  Total_Inventory_Value = 0
  for product in Products:
    Total_Sale = product['Price'] * product['Quantity']
    Total_Inventory_Value += Total_Sale
  return Total_Inventory_Value
  
#Creating a new product list and Calling the function
Product1 = {'Name':'Tomatoes 1kg','Price':781.0,'Quantity':8}
Product2 = {'Name':'Rice 25kg','Price':17258.0,"Quantity":3}
Product3 = {'Name':'Tomato Paste (tin)','Price':378.0,"Quantity":16}
Product4 = {'Name':'Detergent 1kg','Price':1835.0,"Quantity":4}
Product5 = {'Name':'Beans (Red)','Price':899.0	,"Quantity":9}

Products = [Product1, Product2, Product3, Product4, Product5]
print(f"The Total Inventory Value is:{calculate_inventory_value(Products)}FCFA")

```

*args and *kwargs

```python
def add(*numbers):
    total = 0
    for number in numbers:
        total += number
    return total    
add(2,3)
        
```

```python
add(1,2,3,4,5)
```

```python
def marks(**subjects):
    for subject, score in subjects.items():
        print(f"{subject}:{score}")
marks(Chemistry = 70, physics = 50)        
```

```python
marks(math = 40, English = 58 , French = 54, Biology = 75, Geography = 70)
```

*args allows us to input multiple entries and creates a tuple while **kwargs helps us to input key-value pairs, and creates a dictionary

map() functions

```python
#Applying a percentage change to each price using map() (90%)
prices = [34928.0, 685.0, 297.0, 591.0, 915.0]

Changed_Prices = list(map(lambda x: x * 0.9, prices))
print(Changed_Prices )

#Comparing it with a for loop that does thesame process
for price in prices:
    print(price*0.9)
```

filter()

```python
Low_stock = list(filter(lambda x: 0 < x["Quantity"] < 11 , Products ))
print(Low_stock)
```

Combining filteer and map

```python
Well_Stocked = list(map(lambda x: x["Name"], filter(lambda y: y["Quantity"] > 10, Products)))
print(Well_Stocked)

#Separating to see if i get thesame output.
well_stocked = list(filter(lambda d: d["Quantity"] > 10, Products ))
Well_Stocked_Names = list(map(lambda b: b["Name"], well_stocked))
print(Well_Stocked_Names)
```

map does theame role as .apply() ie run a particular function on every item eg list(map(lambda x, x*2, list)) is same as df['col'].apply(lambda x: x*2). filter() keeps items where a condition is true. list(filter(lambda x: x > 10, list)) is thesame as df[df['col']> 10]
