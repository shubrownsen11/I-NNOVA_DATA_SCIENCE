A list of 5 Product dictionaries

```python
Product1 = {'Name':'Salt 1kg','Price':297.0,'Quantity':11}
Product2 = {'Name':'Rice 50kg','Price':34928.0,"Quantity":1}
Product3 = {'Name':'Onions 1kg','Price':685.0,"Quantity":12}
Product4 = {'Name':'Bread (loaf)','Price':591.0,"Quantity":13}
Product5 = {'Name':'Beans (White)','Price':915.0,"Quantity":6}

Products = [Product1, Product2, Product3, Product4, Product5]
Products
```

Its visible that each dictionary is a row, and each key is a column

Using a for loop to print all the contents of the Products list

```python
for product in Products:
    print(f"{product['Name']} costs {product['Price']} and we have {product['Quantity']} in stock. \n")
```

Calculating the total inventory value

```python
Total_Inventory_Value = 0
for product in Products:
    Total_Sale = product['Price'] * product['Quantity']
    Total_Inventory_Value += Total_Sale
print(f"The total Inventory value is {Total_Inventory_Value}FCFA")    
```
