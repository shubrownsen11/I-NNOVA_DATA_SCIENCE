1.comparison operators

```python
print(1==2)
print(4==4)
print(1!=5)
print(3>5)
print(3>=5)
print(3<20)
print(5<=13)
```

2.Basic If/Else

```python
Account_balance  = 150000
if (Account_balance  > 500000):
    print(" Rich")
else:
    print(" Poor") 
   
```

3.Elif

```python
Account_balance  = 150000
if (Account_balance  > 500000):
    print(" Rich")
elif (Account_balance > 250000):
    print("Average") 
else:
    print("poor")
   
```

3.Logical operators

```python
Age = 25
if (Age>18 and Age < 40):
    print ("Can apply for job")

country = "Nigeria"
if (country =="USA" or Age > 20):
    print("eligible")
```

4.Stock status logic

```python
Stock_Quantity = 12
if (Stock_Quantity == 0):
    print("Out of stock")
elif(0 < Stock_Quantity <= 10):
    print("Low Stock")
else:
    print("Well Stocked")   
```

6.Testing against sample quantities

```python
Stock_Quantity = [ 0, 5, 6, 10, 11]
for quantity in Stock_Quantity :
    if (quantity == 0):
      print(f"{quantity}:Out of stock")
    elif(0 < quantity <= 10):
      print(f"{quantity}:Low Stock")
    else:
      print(f"{quantity}:Well Stocked")   
    
    
```
