```python
#Using the 5-product list of dictionaries already built in an earlier lesson (name, price, quantity), writing that data out to a CSV file with a proper header row
Product1 = {'Name':'Salt 1kg','Price':297.0,'Quantity':11}
Product2 = {'Name':'Rice 50kg','Price':34928.0,"Quantity":1}
Product3 = {'Name':'Onions 1kg','Price':685.0,"Quantity":12}
Product4 = {'Name':'Bread (loaf)','Price':591.0,"Quantity":13}
Product5 = {'Name':'Beans (White)','Price':915.0,"Quantity":6}

Products = [Product1, Product2, Product3, Product4, Product5]

fieldnames = list(Products[0].keys())
with open('products.csv','w',encoding='utf-8') as f:
    f.write(",".join(fieldnames) + "\n")
    for row in Products :
        f.write(",".join(str(row.get(key, "")) for key in fieldnames) + "\n")
```

```python
#Reading the CSV file back into python without using pandas as a list of dictionaries
with open('products.csv', 'r', encoding='utf-8') as f:
    lines = f.read().splitlines()
    header = lines[0].split(',')
    Products = []
    for line in lines[1:]:
        if line.strip():
            values = line.split(',')
            Products.append(dict(zip(header, values)))

print(Products)
```

Every column imports as a string. So ill have to convert the relevant columns to int, before ill be able to compute the inventory value

```python
types = {"Name": str, "Price": float, "Quantity": int}

Products = []
for line in lines[1:]:
    if line.strip():
        values = line.split(",")
        row = {key: types[key](val) for key, val in zip(header, values)}
        Products.append(row)
```

```python
#Looping through the re-loaded data and printing a formatted inventory report from it, and confirming the output matches.
total_inventory = 0
for product in Products:
    total_sale = product['Price'] * product['Quantity']
    total_inventory += total_sale
print(f"The Tota Inventory Value is: {total_inventory} FCFA")
```

Step 6 — Combining Everything: Handling a Missing File Gracefully

```python
def read_products(filename):
    header = []
    data = []
    try:
        with open(filename, "r", encoding="utf-8") as f:
            lines = f.read().splitlines()

        header = lines[0].split(",")

        for line_number, line in enumerate(lines[1:], start=2):
            try:
                values = line.split(",")
                if len(values) != len(header):
                    raise ValueError("wrong number of fields")
                row = dict(zip(header, values))
                row["Price"] = float(row["Price"])
                row["Quantity"] = int(row['Quantity'])
                data.append(row)
            except ValueError as e:
                print(f"Row {line_number} has a problem: {e} -> {line}")

        print(f"Loaded {len(data)} valid rows.")
        print(data)

    except FileNotFoundError:
        print(f"Error: the file '{filename}' was not found.")
    finally:
        print("File read attempt complete.")

    return data

read_products("Products.csv")
```
