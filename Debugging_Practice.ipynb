Snippet 4: Total Inventory Value

```python
products = [
 {"name": "Rice 25kg Bag", "price": 15000, "quantity": 12},
 {"name": "Sugar 1kg", "price": 900, "quantity": 50},
 {"name": "Bread Loaf", "price": 700, "quantity": 30},
]
total_value = 0
for product in products:
 total_value = product["price"] * product["quantity"]
print(f"Total inventory value: {total_value} FCFA")
```

```python
products = [
    {"name": "Rice 25kg Bag","price": 15000,"quantity": 12},
    {"name": "Sugar 1kg ","price": 900,"quantity": 50},
    {"name": "Bread Loaf","price": 700,"quantity": 30}
]

total_value = 0
for product in products:
    total_value += product["price"] * product["quantity"]

print(f"Total Inventory Value: {total_value} FCFA")
```

Snippet 4 note:
After running the code i discovered that the code printed just the result of its last computation. The error in the logic was that the total_value wasnt adding up or incrementing.So as the code kept looping, it kept assigning a new value to the total_value variable, rather than incrementing it. I changed the "=" to "+=" so it could increment, and that fixed the bug. Moving forward ill always have an expexted outcome or range subconsciously, so that even if the code doesnt throw an error, i can compare the output and judge its correctness
